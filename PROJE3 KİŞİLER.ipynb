```python
def kısı_oluştur(isim, **özellikler):
   
    kisi_bilgisi = {"isim": isim}
    kisi_bilgisi.update(özellikler)
    print("Kişi oluşturuldu.")
   
    print(f"İsim: {kisi_bilgisi['isim']}")

   
    if "yas" in kisi_bilgisi:
        print(f"Yaş: {kisi_bilgisi['yas']}") 

    # Girinti ve sözlük anahtar erişimi düzeltildi.
    if "meslek" in kisi_bilgisi:
        print(f"Meslek: {kisi_bilgisi['meslek']}") 
    
    
    bilinen_anahtarlar = ["isim", "yas", "meslek", "sehir", "telefon", "email"]
    diger_ozellikler = {k: v for k, v in kisi_bilgisi.items() if k not in bilinen_anahtarlar}
    
  
    if diger_ozellikler:
        print(f"Diğer özellikler: {diger_ozellikler}")
    
    return kisi_bilgisi


kisi1 = kısı_oluştur("Ayşe", yas=30, meslek="Mühendis", sehir="Ankara", hobi="Kitap Okumak")
print("\nOluşturulan kişi bilgisi:", kisi1)
```
